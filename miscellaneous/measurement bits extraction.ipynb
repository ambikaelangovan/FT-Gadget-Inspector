```python
import cirq
```

```python
#one qubit
q = cirq.LineQubit(0)

circuit = cirq.Circuit(cirq.H(q), cirq.measure(q, key='m'))

simulator = cirq.Simulator()
result = simulator.run(circuit, repetitions=1)

print(result)
```

```python
bit = result.measurements['m'][0][0]

print(bit)
print(result.measurements)
```

```python
#two qubits

q0, q1 = cirq.LineQubit.range(2)

circuit = cirq.Circuit(cirq.H(q0), cirq.X(q1), cirq.measure(q0, q1, key='m'))

simulator = cirq.Simulator()
result = simulator.run(circuit, repetitions=1)

print(result)
print(result.measurements)
```

```python
bits = result.measurements['m'][0]

print(bits)
print(result.measurements)
```
